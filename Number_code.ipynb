```python
#check odd/even num from 1 to51
for  num in range(1,51):
    if(num%2==0):
        print('f{num} is even ')
    else:
        print(f'{num} is odd')
```

```python
#multiplication table of 2 and 4
for i in range(1, 11):
    print(f"2 x {i} = {2 * i} \t 4 x {i} = {4 * i}")
```

```python
#sum of evem numbers from 1 to 50
even_sum = sum([num for num in range(1, 51) if num % 2 == 0])
print(f"Sum of even numbers from 1 to 50 is: {even_sum}")
```

```python
#prime num from 1 tp 50
for num in range(2, 51):
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(f"{num} is a Prime Number")
```
